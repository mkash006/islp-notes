```python
import ISLP as islp
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
```

# Numpy basics and vectorized operations

```python
# numpy basics
x=np.random.normal(size=50)
x= x.reshape(5,10)
print("randomly sampled matrix",x, "\n", "dimensions", x.ndim, "\n","type", x.dtype)
y= np.array([1,23,4,4,5,6,7,12,8,10])
print(y.ndim)
```

```python
# useful vectorized and matrix operations
mean = np.mean(x[:,[3,4]])
print("mean for \n", x[:, [3,4]], "is \n" , mean)

np.corrcoef(x[[1,2]],x[[3,4]])

# setting seed
seed = np.random.default_rng(1020)
x1=seed.normal(scale = 5, size=2, loc = 0)
seed2 = np.random.default_rng(1020)
x2 = seed2.normal(scale = 5, size=2, loc = 0)
print(x1, x2)
```

# Graphics with matplotlib

```python
# lineplots
fig , ax = plt.subplots(figsize=(8,8))
x = np.linspace(1,30,2)
y = np.linspace(31,60,2)
ax.plot(x,y)
# scatter plots
fig, ax = plt.subplots (figsize=(8,8))
x = np.random.normal(scale = 10, loc = 12, size = 1000)
y = np.random.normal(scale = 0.5, loc = 110, size = 1000) + x**0.5
ax.plot(x,y,'o')
# mutipanel plot
fig, axes = plt.subplots(nrows = 2,
                      ncols = 2,
                      figsize = (15,5))
axes[0,0].plot(x,y,'o')
axes[0,1].plot(x,y, linestyle = "dotted")
```
